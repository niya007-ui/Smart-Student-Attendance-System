```python
import pandas as pd

ratings = pd.read_csv('ml-latest-small/ratings.csv')
movies = pd.read_csv('ml-latest-small/movies.csv')

print(ratings.head())
print(movies.head())
```

```python
# print(ratings.shape)
# print(ratings['userId'].count())
# print(ratings['movieId'].count())
# print(movies.shape)
# print(ratings['rating'].value_counts())
merged=ratings.merge(movies,on='movieId')

stats=merged.groupby('title')['rating'].agg(['count','mean'])
# popular_movies=stats[stats['count']>=50]
# print(popular_movies.sort_values('mean',ascending=False).head(10))


c=ratings['rating'].mean()

m=50
stats['weighted']=(stats['count']/(stats['count']+m))*stats['mean']+(m/(m+stats['count']))*c
print(stats.sort_values('weighted',ascending=False).head(10))

# top10=stats.sort_values('count',ascending=False).head(10)
# top10['mean']=top10['mean'].round(2)
# top10['mean']=top10['mean'].round(2)
# print(top10)
```
